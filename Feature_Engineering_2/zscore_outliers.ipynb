```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore
```

```python
data = {
    "value" : [10, 23, 43, 65, 43, 23, 15, 54, 100, 32]
}
df = pd.DataFrame(data)
df
```

### Z-Score Method

```python
df['z-score'] = zscore(df['value'])
```

```python
df
```

```python
threshold = 1.5
```

```python
df['outliers'] = np.abs(df['z-score']) > threshold
```

```python
df
```

```python
sns.boxplot(y=df['value'])
plt.title("Before Removing Outlier")
```

```python
df_clean = df[df['outliers'] == False]
```

```python
df_clean
```

```python
sns.boxplot(y=df_clean['value'])
plt.title("After Removing Outlier")
```
