```python
import pandas as pd 

df = pd.read_csv("tested.csv")

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1.Catagorical Univariate Countplot 

    A statistical method that examines a single variable in a data set to describe it and find patterns

```python
sns.countplot(x='Sex', data=df)
```

```python
df['Sex'].value_counts()
```

```python
sns.countplot(x='Survived', data=df)
```

```python
sns.countplot(x='Pclass', data=df)
df['Pclass'].value_counts()
```

### 2.Piechart

```python
df['Pclass'].value_counts().plot(kind='pie', autopct='%.0f')
```

### 1.Numerical Univariate

### 2. Bivariate 
    Bivariate analysis refers to the analysis of two variables to determine relationships between them

#### 1.Numerical - Numerical
     1.Scatter Plot

```python
tips = pd.read_csv('tip.csv')

tips.head()
```

```python
sns.scatterplot(x=tips['total_bill'], y=tips['tip'], hue=tips['sex'])
```

```python
tips_groupby = tips.groupby('sex')['tip'].sum().reset_index()
tips_groupby.head()
```

```python
sns.lineplot(x=tips_groupby['sex'], y=tips_groupby['tip'])
plt.show()
```
