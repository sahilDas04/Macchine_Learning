```python
### Feature Selection Based on correlation
X_train.corr()
```

```python
### Check for multicollinearity
plt.figure(figsize=(12, 10))
corr = X_train.corr()
sns.heatmap(corr, annot=True)
```

```python
def correlation(dataset, threshold):
    col_corr = set()
    corr_matrix = dataset.corr()
    for i in range(len(corr_matrix.columns)):
        for j in range(i):
            if abs(corr_matrix.iloc[i, j]) > threshold:
                colname = corr_matrix.columns[i]
                col_corr.add(colname)
    return col_corr
```

```python
### Threshold-->Domain Expert
corr_features = correlation(X_train, 0.80)
```

```python
### Check which features are identified for removal
print("Features to drop:", corr_features)
print("\nCurrent X_train columns:", X_train.columns.tolist())
print("\nFeatures in corr_features that exist in X_train:", [f for f in corr_features if f in X_train.columns])
```

```python
### Drop features when correlation is more than 0.80
X_train.drop(corr_features, axis=1, inplace=True)
X_test.drop(corr_features, axis=1, inplace=True)
```
