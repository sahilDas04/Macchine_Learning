### Assignment 6: List Methods

Create a list of random numbers and sort it in ascending and descending order. Remove the duplicates from the list and print the modified list.

```python
import random

random_numbers = []

for i in range(20):
    number = random.randint(1, 20)
    random_numbers.append(number)


print(f"Original list: {random_numbers}")

random_numbers.sort()
print(f"Sorted List : {random_numbers}")

random_numbers.reverse()
print(f"Reversed List : {random_numbers}")

unique_number = list(set(random_numbers))
print(f"Removed Duplicate List : {unique_number}")
```

### Assignment 9: Matrix Transposition

Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.

```python
def transpose(matrix):
    transposed = [[matrix[i][j] for i in range(len(matrix))] for j in range(len(matrix[0]))]
    return transposed

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

transpose = transpose(matrix)

for i in matrix:
    print(i)

print("Transepose matrix : ")

for i in transpose:
    print(i)
```

### Assignment 10: Flattening a Nested List

Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.

```python
def flat(nes_lis):
    flat_list = [item for sub in nes_lis for item in sub]
    return flat_list

nes_lis = [
    [1, 2, 3],
    [4, 5, 6], 
    [7, 8, 9]
]

print(nes_lis)

print(flat(nes_lis))
```

### Assignment 13: List Reversal

Write a function that takes a list and returns a new list with the elements in reverse order. Print the original and reversed lists.

```python
def reversed(lst):
    return lst[::-1]

lst = list(range(1, 11))
print(f"The original list : {lst}")
print(f"The reversed list : {reversed(lst)}")
```

### Assignment 14: List Rotation

Write a function that rotates a list by n positions. Print the original and rotated lists.

```python
def rotation(lst, n):
    return lst[n:] + lst[:n]

lst = list(range(1, 6))
print(f"The original list : {lst}")

x = int(input("Enter the number : "))
print(f"The rotate list : {rotation(lst, x)}")
```

### Assignment 15: List Intersection

Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.

```python
def inter_section(lst1, lst2):
    return [x for x in lst1 if x in lst2]

list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

print(f"The Intersection list : {inter_section(list1, list2)}")
```
