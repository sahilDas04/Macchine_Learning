### Assignment 7: Handling Transactions

1. Write a Python function to insert multiple employees into the `employees` table in a single transaction. Ensure that if any insertion fails, none of the insertions are committed.

```python
import sqlite3

def insert_many_employee(employees):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    try:
        cursor.executemany('''
            INSERT INTO employees(id, name, age, department)
            VALUES(?, ?, ?, ?)
        ''', employees)  # Pass employees directly
        conn.commit()
        print("All employees inserted successfully.")
    except Exception as e:
        conn.rollback()
        print("Error occurred, transaction rolled back.")
        print(e)
    finally:
        conn.close()

# Test the function with valid and invalid data
employees = [
    (6, 'Frank', 40, 'Finance'),
    (7, 'Grace', 29, 'Engineering'),
    (8, 'Hannah', 35, 'Marketing'),
    (9, 'Ivan', 38, 'Sales'),
]

insert_many_employee(employees)
```

2. Write a Python function to update the age of multiple employees in a single transaction. Ensure that if any update fails, none of the updates are committed.

```python
def update_age(update):
   conn = sqlite3.connect('test.db')
   cursor = conn.cursor()
   try:
      cursor.executemany('''
         UPDATE employees
         SET age = ?
         WHERE id = ?
      ''', update)
      conn.commit()
      print("All employ age are updated!!")
   except Exception as e:
      conn.rollback()
      print("Error occurred, transaction rolled back.")
      print(e)
   finally:
      conn.close()
updates = [
   (32, 6),
   (26, 7),
   (33, 8),
   (23, 9)
]
update_age(updates)   
```

### Assignment 8: Creating Relationships

1. Create a new table named `departments` with columns `id` (integer) and `name` (text).

```python
def new_table():
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS departments(
            id INTEGER PRIMARY KEY,
            name TEXT NOT NULL
        )              
    ''')
    conn.commit()
    conn.close()
    print("The department table created successfully!!")

new_table()
```

2. Modify the `employees` table to include a foreign key referencing the `id` column in the `departments` table.

```python
def add_foreign_key():
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    
    try:
        cursor.execute('PRAGMA foreign_keys=off;')
        cursor.execute('BEGIN TRANSACTION;')
        cursor.execute('ALTER TABLE employees RENAME TO old_employees;')
        cursor.execute('''
            CREATE TABLE employees (
                id INTEGER PRIMARY KEY,
                name TEXT NOT NULL,
                age INTEGER,
                department TEXT,
                department_id INTEGER,
                FOREIGN KEY(department_id) REFERENCES departments(id)
            );
        ''')
        cursor.execute('''
            INSERT INTO employees (id, name, age, department)
            SELECT id, name, age, department FROM old_employees;
        ''')
        cursor.execute('DROP TABLE old_employees;')
        conn.commit()
    except Exception as e:
        conn.rollback()
        print(f"Error: {e}")
    finally:
        cursor.execute('PRAGMA foreign_keys=on;')
        conn.close()
    print("Table 'employees' modified successfully.")
add_foreign_key()
```

3. Write a Python function to insert data into both the `departments` and `employees` tables, ensuring referential integrity.

```python
import sqlite3

def insert_employee_department(depart_id, employee_id, name, age, depart_name, depart):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    try:
        cursor.execute('''
            INSERT INTO departments(id, name)
            VALUES(?, ?)
        ''', (depart_id, depart_name))
        cursor.execute('''
            INSERT INTO employees(id, name, age, department, department_id)
            VALUES(?, ?, ?, ?, ?)
        ''', (employee_id, name, age, depart, depart_id))
        
        conn.commit()
        print("Department and employee inserted successfully.")
    except Exception as e:
        conn.rollback()
        print("Error occurred, transaction rolled back.")
        print(e)
    finally:
        conn.close()
insert_employee_department(1, 10, 'Zara', 28, 'Finance', 'Finance')
```
