## Sql and Sqlite

#### Sqlite
 SQLite is software which provides relational database management system. SQLite lightweight in terms of setup, database administration, and required resources. SQLite has features like self-contained, server-less, zero-configuration, etc.

#### Sql(Structured Query Language)
 SQL stands for Structured Query Language. SQL can access, created and manage databases.

```python
import sqlite3
```

```python
## Connect the database 
connection = sqlite3.connect('Example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
cursor.execute('''
    create table if not exists employee(
        id integer primary key,
        name text not null,
        age integer,
        department text
    )
''')
connection.commit()
```

```python
cursor.execute('''
    select * from employee;
''')
```

```python
cursor.execute('''
    insert into employee(name, age, department) values('Sahil', 20, 'Data Science')
''')

cursor.execute('''
    insert into employee(name, age, department) values('Manish', 19, 'Web Devlopment')
''')

cursor.execute('''
    insert into employee(name, age, department) values('Pratyush', 25, 'Web Devlopment')
''')

## Commit the changes
connection.commit()
```

```python
cursor.execute('select * from employee')

rows = cursor.fetchall()

for row in rows:
    print(row)
```

#### Update the data on the database 

```python
cursor.execute('''
    update employee
    set age = 19
    where name="Manish"
''')
connection.commit()
```

```python
cursor.execute('select * from employee')

rows = cursor.fetchall()

for row in rows:
    print(row)
```

#### Delete the data from DataBase

```python
cursor.execute('''
    delete from employee
    where name = "Sahil"
''')

connection.commit()
```
