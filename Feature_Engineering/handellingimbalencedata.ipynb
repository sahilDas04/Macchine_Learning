### 1.Upsampling Method

```python
import pandas as pd 
import numpy as np

np.random.seed(123)

n_sample = 1000

class_0_ratio = 0.9
n_class_0 = int(n_sample * class_0_ratio)
n_class_1 = n_sample - n_class_0
```

```python
n_class_0, n_class_1
```

```python
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=1, scale=2, size=n_class_1),
    'feature_2': np.random.normal(loc=1, scale=2, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
class_0, class_1
```

```python
df = pd.concat([class_0, class_1]).reset_index(drop=True)
```

```python
df.head(5)
```

```python
df.tail(5)
```

```python
df['target'].value_counts()
```

```python
df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample

df_minority_upsamled = resample(df_minority, 
                                replace=True, 
                                n_samples=len(df_majority), 
                                random_state=123
                                )
```

```python
df_minority_upsamled.shape
```

```python
df_minority_upsamled.head(5)
```

```python
df_upsampled = pd.concat([df_majority, df_minority_upsamled])
```

```python
df_upsampled['target'].value_counts()
```

### 2.Downsampling Method

```python
import pandas as pd 
import numpy as np

np.random.seed(123)

n_sample = 1000
class_ratio = 0.9
n_class_0 = int(n_sample * class_ratio)
n_class_1 = n_sample - n_class_0
```

```python
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=1, scale=2, size=n_class_1),
    'feature_2': np.random.normal(loc=1, scale=2, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
df = pd.concat([class_0, class_1]).reset_index(drop=True)
```

```python
df.shape
```

```python
df.head(5)
```

```python
df['target'].value_counts()
```

```python
df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample

df_majority_downsampled = resample(df_majority, 
              replace=False,
              n_samples=len(df_minority),
              random_state=123)
```

```python
df_majority_downsampled = pd.concat([df_minority, df_majority_downsampled])
```

```python
df_majority_downsampled['target'].value_counts()
```
