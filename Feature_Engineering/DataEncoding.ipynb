## Data Encoding
     when a categorical data coverted into a numerical data it's called Data Encoding

#### One Hot Encoding
    A "OneHotEncoder" creates a new binary column for each unique category in a categorical feature, representing presence with a "1" and absence with a "0"

```python
import pandas as pd 
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color' : ['red', 'green', 'blue', 'red', 'green', 'blue']
})
```

```python
df
```

```python
encode = OneHotEncoder()
```

```python
encoder = encode.fit_transform(df[['color']]).toarray()
```

```python
encoder
```

```python
encoded_df = pd.DataFrame(encoder, columns=encode.get_feature_names_out(['color']))
```

```python
encoded_df
```

```python
encode.transform([['blue']]).toarray()
```

```python
pd.concat([df, encoded_df], axis=1)
```

#### Label Encoding
    "LabelEncoder" simply assigns a unique numerical value to each category, potentially implying an order even if none exists

```python
from sklearn.preprocessing import LabelEncoder

lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
lbl_encoder.transform(['red'])
```

#### Ordinal Encoding
    When we assing the ranks to the particular value and particular columns then we use ordinal encoder.

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
ordinal_encoder = OrdinalEncoder()
```

```python
df = pd.DataFrame({
    'size' : ['small', 'medium', 'large', 'small', 'medium', 'large']
})
df
```

```python
encoder = OrdinalEncoder(categories=[['small', 'medium', 'large']])
```

```python
encoder.fit_transform(df[['size']])
```

#### Target Guided Ordinal Encoding
    It is a technique which is used to encode categorical variable based on their relationship with the target variable. This encoding technique is usefull when we have categorical variable with a large numbr of unique categories.

```python
import pandas as pd 
```

```python
df = pd.DataFrame({
    'City': ['New York', 'Los Angeles', 'Chicago', 'New York', 'Los Angeles', 'Chicago'],
    'Price': [1000, 850, 750, 900, 800, 700]
})

df
```

```python
mean_price = df.groupby('City')['Price'].mean().to_dict()
```

```python
mean_price
```

```python
df['City_Encoded'] = df['City'].map(mean_price)
```

```python
df
```

```python
df[['Price', 'City_Encoded']]
```
