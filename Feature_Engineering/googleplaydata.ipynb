```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Data Cleaning

```python
df['Reviews'].astype(int)
```

```python
df['Reviews'].unique()
```

```python
df['Reviews'].str.isnumeric().sum()
```

```python
df[~df['Reviews'].str.isnumeric()]
```

```python
df_new = df.copy()
```

```python
df_new.head(2)
```

```python
df_new = df_new.drop(df_new.index[10472])
```

```python
df_new[~df_new['Reviews'].str.isnumeric()]
```

```python
df_new.info()
```

```python
df_new['Reviews'] = df_new['Reviews'].astype(int)
```

```python
df_new.info()
```

```python
df_new['Size'].unique()
```

```python
df_new['Size'].isnull().sum()
```

```python
df_new['Size'] = df_new['Size'].str.replace('M', '000')
df_new['Size'] = df_new['Size'].str.replace('k', '')
df_new['Size'] = df_new['Size'].replace('Varies with device', np.nan)
df_new['Size'] = df_new['Size'].astype(float)
```

```python
df_new['Size']
```

```python
df.head()
```

```python
df_new['Installs'].unique()
```

```python
df_new['Price'].unique()
```

```python
char_to_remove = ['+', ',', '$']
cols_to_remove = ['Installs', 'Price']
for item in char_to_remove:
    for cols in cols_to_remove:
        df_new[cols] = df_new[cols].str.replace(item, '')
```

```python
df_new['Installs'].unique()
```

```python
df_new['Price'].unique()
```

```python
df_new['Installs'] = df_new['Installs'].astype(int)
df_new['Price'] = df_new['Price'].astype(float)
```

```python
df_new.info()
```

```python
df_new['Last Updated'].unique()
```

```python
df_new['Last Updated'] = pd.to_datetime(df_new['Last Updated'])
df_new['Day'] = df_new['Last Updated'].dt.strftime('%d').astype(int)
df_new['Month'] = df_new['Last Updated'].dt.month
df_new['Year'] = df_new['Last Updated'].dt.year
```

```python
df_new.info(1)
```

```python
df_new.head(1)
```

```python
df_new.to_csv('googleplaystore_clean.csv')
```

```python
df_new = df_new.drop_duplicates(subset=['App'], keep='first')
```

```python
df_new.shape
```

```python
numeric_features = [feature for feature in df_new.columns if df_new[feature].dtypes != 'O']
categorical_features = [feature for feature in df_new.columns if df_new[feature].dtypes == 'O']

print('We have {} numerical features. {}'.format(len(numeric_features), numeric_features))
print('\n We have {} categorical features. {}'.format(len(categorical_features), categorical_features))
```

```python
for cols in categorical_features:
    print(df[cols].value_counts(normalize=True)*100)
    print('-------------------------------------------')
```

```python
# Proportion of count data on numeric features
plt.figure(figsize=(15, 15))
plt.suptitle('Univariate Analysis of numerical features', fontsize=20, fontweight='bold', alpha=0.8, y=1.)

for i in range(0, len(numeric_features)):
    plt.subplot(5, 3, i+1)
    sns.kdeplot(x=df_new[numeric_features[i]], shade=True, color='r')
    plt.xlabel(numeric_features[i])
    plt.tight_layout()
```

## Observation
- Rating and Year is left skewed while Installs, Price, Reviews, and Size is Right skewed

```python
# Proportion of count data on numeric features
plt.figure(figsize=(20, 15))
plt.suptitle('Univariate Analysis of categorical features', fontsize=20, fontweight='bold', alpha=0.8, y=1.)
category = ['Type', 'Content Rating']

for i in range(0, len(category)):
    plt.subplot(2, 2, i+1)
    sns.countplot(x=df_new[category[i]], palette='Set2')
    plt.xlabel(category[i])
    plt.xticks(rotation=45)
    plt.tight_layout()
```

### Which category has largest number of installation??

```python
category_install = df_new.groupby('Category')['Installs'].sum().sort_values(ascending=False)
```

```python
category_install
```

```python
plt.figure(figsize=(15, 6))
sns.barplot(x=category_install.index, y=category_install, palette='hls')
plt.title('Total instalation per category')
plt.ylabel('Total Installations')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Observation
- Game, Communication, Tools have largest number of installation in playstore

### Which are the top 5 most installed app in each category??

```python
top_5_apps = df_new.sort_values('Installs', ascending=False).groupby('Category').head(5)
```

```python
top_5_apps
```

```python
plt.figure(figsize=(20, 100))
sns.barplot(x='Installs', y='App', hue='Category', data=top_5_apps, palette='viridis')

plt.title('Top 5 Most Installed Apps in Each Category')
plt.xlabel('Installations')
plt.ylabel('App')
plt.tight_layout()
plt.show()
```
