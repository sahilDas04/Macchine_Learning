```python
import numpy as np
import random
```

```python
n = 20 
list_marks = random.sample(range(1, 100), n)
```

```python
print(list_marks)
```

```python
minimum, Q1, median, Q3, maximum = np.quantile(list_marks, [0, 0.25, 0.50,0.75, 1])
```

```python
minimum, Q1, median, Q3, maximum
```

```python
IQR = Q3 - Q1
IQR
```

```python
lower_fence = Q1 - (1.5*IQR)

lower_fence
```

```python
higher_fence = Q3 + (1.5*IQR)

higher_fence
```
