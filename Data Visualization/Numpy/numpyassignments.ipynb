### Assignment 4: Statistical Operations

1. Create a NumPy array of shape (5, 5) filled with random integers. Compute the mean, median, standard deviation, and variance of the array.

```python
arr = np.random.randint(1, 9, size=(5, 5))
print(arr)

print(f"The mean of the data : {np.mean(arr)}")
print(f"The standard deviation of the data : {np.std(arr)}")
print(f"The median of the data : {np.median(arr)}")
print(f"The variance of the data : {np.var(arr)}")
```

2. Create a NumPy array of shape (3, 3) with values from 1 to 9. Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).

```python
data = np.arange(1, 10).reshape((3, 3))
print(data)

data_mean = np.mean(data)
data_std = np.std(data)

normi = (data - data_mean)/data_std
print("\nNormalize Data : \n")
print(normi)
```

### Assignment 10: Masked Arrays

1. Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements.

```python
import numpy.ma as ma
arr = np.random.randint(1, 17, size=(4, 4))
mask_arr = ma.masked_greater(arr, 10)
print("Original array : ")
print(arr)
print("Maked array : ")
print(mask_arr)

unmask_sum = mask_arr.sum()
print(f"The sum of unmasked arr : {unmask_sum}")
```

2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.

```python
import numpy.ma as ma
arr = np.random.randint(1, 17, size=(3, 3))
mask_arr = ma.masked_array(arr, mask=np.eye(3, dtype=bool))
print("Original matrix")
print(arr)
print("Masked Matrix : ")
print(mask_arr)

arr_mean = arr.mean()
masked = mask_arr.filled(arr_mean)
print("Modified Array : ")
print(masked)
```
