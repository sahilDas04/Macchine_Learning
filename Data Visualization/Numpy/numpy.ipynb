#### Vectorized operaton

```python
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([10, 20, 30, 40, 50])

print(f"Addition : {arr1+arr2}")
print(f"Subtraction : {arr1-arr2}")
print(f"Multiplication : {arr1*arr2}")
print(f"Division : {arr1/arr2}")
```

```python
arr = np.arange(1, 13).reshape((3, 4))
print(arr)
```

```python
print(arr[0][0])
```

```python
print(arr[1:, 2:])
```

```python
print(arr[0:2, 2:])
```

```python
arr[1:] = 100
print(arr)
```
