### Seaborn

Seaborn is a python visualization library based on matplot that provides a high level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualization just few line of code.

```python
import pandas as pd

tips = pd.read_csv('tip.csv')
```

```python
import seaborn as sns
```

##### Basic Ploting with Seaborn

```python
tips = sns.load_dataset('tips')
```

```python
tips.head(5)
```

```python
import matplotlib.pyplot as plt
```

```python
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title("Scatter plot of total bill vs tip")
```

```python
sns.lineplot(x='size', y='total_bill', data=tips)
plt.title("Line plot of total bill vs size")
```

##### Categorical polt

```python
## Bar Plot
sns.barplot(x='day', y='total_bill', data=tips)
plt.title("Bar plot on Day and Total bill")
```

```python
## Box Plot
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title("Box plot on Day and Total bill")
```

```python
## Violin plot
sns.violinplot(x='day', y='total_bill', data=tips)
plt.title("Violin plot on Day and Total bill")
```

```python
## Histogram 
sns.histplot(tips['total_bill'], bins=10, kde=True)
```

```python
### Kde PLot
sns.kdeplot(tips['total_bill'], fill=True)
```

```python
### Pairplot
sns.pairplot(tips)
```

```python
corr = tips[['total_bill', 'tip', 'size']].corr()
corr
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
```

```python
sale_df = pd.read_csv('dataset.csv')
sale_df.head(5)
```

```python
plt.figure(figsize=(20, 12))
sns.barplot(x='Product line', y='Total', data=sale_df, estimator=sum)
plt.title('Total Sale by product')
plt.xlabel('Product Name')
plt.ylabel('Total Revenue')
plt.show()
```

```python
plt.figure(figsize=(20, 12))
sns.barplot(x='City', y='Total', data=sale_df, estimator=sum)
plt.title('Total Revenue by City')
plt.xlabel('City Name')
plt.ylabel('Total Revenue')
plt.show()
```
