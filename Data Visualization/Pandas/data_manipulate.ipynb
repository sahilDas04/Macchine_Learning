# Data Manipulation and analysis using pandas and numpy

```python
import pandas as pd
import numpy as np

df = pd.read_csv('sale_data.csv')
df.head(5)
```

```python
df.tail(5)
```

```python
df.describe()
```

```python
df.dtypes
```

```python
# Handling missing values

df.isnull().any(axis=1)
```

```python
df.isnull().any()
```

```python
df.isnull().sum()
```

```python
## If there is a missing value we can also replace by 0 or anything you want 
## df_filled = df.fillna(0) we have to copy that data value

## Filling missing value the the mean of the columns value
## df['variable or column name'] = df['columns name'].fillna(df['same columns name'].mean()) ;  This will create a new columns and fill the mean of each columns also the missing value columns
df['Mean'] = df['Total'].fillna(df['Total'].mean())
df
df.drop('Mean of each col',axis=1, inplace=True)
```

```python
df.head(10)
```

```python
# Renameing the columns 

df = df.rename(columns={'Invoice ID' : 'ID'})
df.head(5)
```

```python
# changing datatype

df['New'] = df['Total'].fillna(df['Total'].mean()).astype(int) # Used fillna because there might be a missing value
df.head(5)
```

```python
df['New total'] = df['New'].apply(lambda x:x*2)
df.head(5)
```

#### Data aggregrating and grouping

```python
## GroupBy using 

group_mean = df.groupby('Product line')['Total'].mean()
group_mean
```

```python
group_sum = df.groupby(['Product line', 'City'])['Total'].sum()
group_sum
```

```python
## multiple aggregrate function

group_agg = df.groupby('City')['Total'].agg(['mean', 'sum', 'count'])
group_agg
```

##### Merging and Joining DataFrame 

```python
df1 = pd.DataFrame({'key' : ['A', 'B', 'C'], 'value1' : [1, 2, 3]})
df2 = pd.DataFrame({'key' : ['A', 'B', 'C'], 'value2' : [1, 2, 3]})
```

```python
df1
```

```python
df2
```

```python
merged_data = pd.merge(df1, df2, on='key', how='inner')
print(merged_data)
```
