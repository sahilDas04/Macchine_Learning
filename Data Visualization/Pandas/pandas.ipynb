## Pandas
 #### 1. Series

 A pandas series is a one dimensional array like object that can hold any data type. It similar to a column in a table.

```python
import pandas as pd

data = [1, 2, 3, 4, 5, 6]
series = pd.Series(data)
print(series)
print(type(series))
```

```python
data = {'a':1, 'b':2, 'c':3, 'd':4, 'e':5}

series_dict = pd.Series(data)
print(series_dict)
print(type(series_dict))
```

```python
data = [1, 2, 3, 4]
idx = ['a', 'b', 'c', 'd']

serie = pd.Series(data, index=idx)
print(serie)
```

## Dataframe
DataFrame is a 2-dimensional labeled data structure with columns of potentially different types.

```python
import pandas as pd
data = {
    'Name' : ['Sahil', 'Manish', 'pratyush'],
    'Age' : [20, 19, 25],
    'city' : ['Kolkata', 'bihar', 'punjab']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

##### Create a Dataframe using Dictionaries

```python
data = [
    {'Name' : 'Sahil', 'Age' : 20, 'city' : 'Kolkata'},
    {'Name' : 'Mani', 'Age' : 19, 'city' : 'bihar'},
    {'Name' : 'pratyush', 'Age' : 25, 'city' : 'punjab'},
]

df = pd.DataFrame(data)
print(df)
```
