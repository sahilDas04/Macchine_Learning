### Assignment 3: Data Cleaning

1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

```python
df = pd.DataFrame(np.random.randint(1, 9, size=(5, 3)), columns=['A', 'B', 'C'])
print("Original DataFrame : ")
print(df)

df.iloc[0, 1] = np.nan
df.iloc[2, 2] = np.nan
df.iloc[4, 0] = np.nan
df.iloc[3, 0] = np.nan
df.iloc[1, 0] = np.nan
print("NaN values with dataframe : ")
print(df)

df.fillna(df.mean(), inplace=True)
print("Modified Dataframe : ")
print(df)
```

### Assignment 5: Merging DataFrames

1. Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.


```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame({'key' : ['A', 'B', 'C'], 'value1' : np.random.randint(1, 10, size=3)})
df2 = pd.DataFrame({'key' : ['A', 'B', 'C'], 'value2' : np.random.randint(1, 10, size=3)})

print("1st Dataframe : ")
print(df1)
print("2nd Dataframe : ")
print(df2)

merge_data = pd.merge(df1, df2, on='key')
print('Modified Dataframe : ')
print(merge_data)
```

2. Create two Pandas DataFrames with different columns. Concatenate the DataFrames along the rows and along the columns.

```python
df1 = pd.DataFrame({'key' : ['A', 'B', 'C'], 'value1' : np.random.randint(1, 10, size=3)})
df2 = pd.DataFrame({'key' : ['A', 'B', 'C'], 'value2' : np.random.randint(1, 10, size=3)})

print("1st Dataframe : ")
print(df1)
print("2nd Dataframe : ")
print(df2)

concate1 = pd.concat([df1, df2], axis=0)
concate2 = pd.concat([df1, df2], axis=1)
print("The row wise concatenation : ")
print(concate1)
print("The columns wise concatenation : ")
print(concate2)
```
